```python
# Load two images
img1 = cv2.imread('image1.jpg')
img2 = cv2.imread('image2.jpg')

# Check whether images are loaded
if img1 is None or img2 is None:
    raise FileNotFoundError("Please check the image file names and paths.")

print("Both images loaded successfully!")
```

```python
# Resize both images to 500 x 500
img1 = cv2.resize(img1, (500, 500))
img2 = cv2.resize(img2, (500, 500))

print("Images resized successfully!")
```

```python
# Blend two images
blended = cv2.addWeighted(img1, 0.5, img2, 0.5, 0)

print("Image blending completed!")
```

```python
# Convert BGR to RGB for displaying with Matplotlib
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
blended_rgb = cv2.cvtColor(blended, cv2.COLOR_BGR2RGB)

# Display images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img1_rgb)
plt.title("Original Image 1")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img2_rgb)
plt.title("Original Image 2")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(blended_rgb)
plt.title("Blended Image")
plt.axis('off')

plt.show()
```

```python
# Save original and blended images
cv2.imwrite('original_image1.jpg', img1)
cv2.imwrite('original_image2.jpg', img2)
cv2.imwrite('blended_image.jpg', blended)

print("All images saved successfully!")
```
