```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

data = {
    'f1': np.random.randint(1, 100, 10),
    'f2': np.random.randint(1, 100, 10),
    'f3': np.random.randint(1, 100, 10),
    'f4': np.random.randint(1, 100, 10)
}
```

```python
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
```

```python
plt.scatter(df['f2'], df['f3'], color='blue')

plt.xlabel('f2')
plt.ylabel('f3')
plt.title('Scatter Plot of f2 vs f3')
plt.grid(True)
plt.show()
```

```python
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

print("\nDataFrame with f5 (Mean):")
print(df)
```

```python
df.to_excel('output.xlsx', index=False)

print("\nData successfully exported to output.xlsx")
```
