# Exercise 5: Interactive AI Prototyping with Gradio

**Project:** StudyBuddy AI — Chat Lab

**Objective:** Build and launch a lightweight browser-based conversational interface using Python and Gradio.

This prototype uses `gr.Blocks` to create a chat area, message box, send and clear controls, a welcome description, and four sample prompts. It runs in Demo Mode without requiring an API key, making it easy to launch and evaluate in a Jupyter environment.

**Note:** The response function is a lightweight local demo engine. It demonstrates the conversational interface and chat flow; it is not presented as a pretrained large language model.

## 2. Import required library

```python
import gradio as gr
```

## 3. Define the chatbot response function

```python
def respond(message, history):
    text = message.lower().strip()

    if any(word in text for word in ["hello", "hi", "hey"]):
        return "Hello! Welcome to StudyBuddy AI. What would you like to learn today?"
    elif "artificial intelligence" in text or text == "ai":
        return ("Artificial Intelligence (AI) creates computer systems that perform "
                "tasks such as learning, reasoning, pattern recognition, and language "
                "understanding. Examples include voice assistants and recommendation systems.")
    elif "machine learning" in text:
        return ("Machine Learning is a branch of AI in which systems learn patterns "
                "from data. For example, an email filter can learn to identify spam.")
    elif "prompt" in text:
        return ("A prompt is an instruction given to an AI system. A strong prompt "
                "defines the task, context, expected format, and constraints.")
    elif "gradio" in text:
        return ("Gradio is a Python library for creating web interfaces for Python "
                "functions and machine-learning models.")
    else:
        return ("This prototype is running in Demo Mode with a small local response set. "
                "Try asking about AI, Machine Learning, prompt engineering, or Gradio.")
```

## 4. Design the Gradio interface

```python
with gr.Blocks(theme=gr.themes.Soft(), title="StudyBuddy AI | Chat Lab") as demo:
    gr.Markdown(
        "# ✨ StudyBuddy AI — Chat Lab\n"
        "**A mini conversational AI prototype built with Python and Gradio.**\n\n"
        "Ask a question or select a starter prompt below. This lightweight version "
        "runs in Demo Mode and does not require an API key."
    )

    chatbot = gr.Chatbot(label="Conversation", height=380, bubble_full_width=False)
    message = gr.Textbox(
        label="Your message",
        placeholder="Ask a question...",
        lines=2
    )

    with gr.Row():
        send = gr.Button("Send message", variant="primary")
        clear = gr.Button("Clear chat")

    gr.Examples(
        examples=[
            "What is Artificial Intelligence?",
            "Explain Machine Learning with a daily-life example.",
            "What makes a good AI prompt?",
            "How does Gradio help build AI prototypes?"
        ],
        inputs=message,
        label="Sample questions"
    )

    def chat(message_text, history):
        answer = respond(message_text, history)
        history = history or []
        history = history + [
            {"role": "user", "content": message_text},
            {"role": "assistant", "content": answer}
        ]
        return history, ""

    send.click(chat, inputs=[message, chatbot], outputs=[chatbot, message])
    message.submit(chat, inputs=[message, chatbot], outputs=[chatbot, message])
    clear.click(lambda: ([], ""), outputs=[chatbot, message])
```

## 5. Launch the chatbot

```python
demo.launch(share=False)
```
