# Exercise 3: Generative AI — Prompt Engineering for Video Synthesis

## 1. Selected AI Video-Generation Tool
**Selected tool:** Kling AI

Kling AI is selected for the image-to-video prompt. The prompt below is written for the three reference visuals. A 3-second preview clip is also included with this submission; it was assembled from the supplied images with smooth transitions and a newly created instrumental soundtrack.

## 2. Objective
To formulate a structured prompt for a 3-second video using the two source images and their blended image from Exercise 2.

## 3. Reference Images
- **Source Image 1:** A moonlit fantasy garden with a glowing full moon, flowering tree, mist, and green foliage.
- **Source Image 2:** A peaceful lake with two white swans and pink cherry-blossom trees.
- **Blended Image:** A combined fantasy landscape that transitions from the cherry-blossom lake scene into the moonlit garden.

## 4. Exact Prompt Text

> Create a 3-second vertical 9:16 fantasy nature video using the three supplied reference images. Begin with the moonlit enchanted garden: a bright full moon behind a soft cloud, a flowering pink tree, glowing stars, mist, and lush green plants. Transition smoothly to the peaceful cherry-blossom lake with two elegant white swans floating on still water beneath pink flowering trees. Then transition into the blended reference image, combining the pink cherry-blossom landscape with the glowing, misty green garden. Preserve the key details and colors of each reference image. Add subtle natural movement: gently drifting mist, a slow cloud movement, a few falling petals, and slight ripples on the lake. Use a calm, magical atmosphere, soft cinematic lighting, and a stable camera. Avoid sudden cuts, flicker, distorted trees, extra animals, text, or logos.

## 5. Camera and Transition Instructions

- **Duration:** 3 seconds
- **Aspect ratio:** 9:16 (vertical)
- **Resolution:** 1080 × 1920
- **Frame rate:** 30 fps
- **Opening (0.0–1.0 sec):** Show the moonlit fantasy garden.
- **Transition 1:** Use a smooth cross-dissolve into the cherry-blossom lake and swans.
- **Middle (approximately 1.0–2.0 sec):** Show the swans and pink cherry-blossom trees with a subtle slow push-in.
- **Transition 2:** Cross-dissolve into the blended landscape.
- **Ending (approximately 2.0–3.0 sec):** Hold the blended image with a gentle camera push-in.
- **Audio:** Soft, original ambient instrumental music; no dialogue or lyrics.
- **Visual style:** Dreamlike, peaceful, cinematic, colorful, and nature-focused.
